```python
#@title 💚 WALL•ET | Configuración, Login y Bienvenida

import ipywidgets as widgets
from IPython.display import display, clear_output
import json
import os

def formato_dinero(cantidad):
    return f"${cantidad:,.0f}".replace(",", ".")

pantalla = widgets.Output()

# Variables globales de memoria
usuario_actual = ""
datos_usuario = {
    "ingreso_mensual": 0,
    "metas": [],
    "presupuestos": {},
    "ingresos": [],
    "gastos": []
}

# ==========================================
# FUNCIONES DE GUARDADO (MEMORIA JSON)
# ==========================================
def cargar_perfil(nombre):
    global usuario_actual, datos_usuario
    usuario_actual = nombre.strip().lower()
    archivo = f"wallet_{usuario_actual}.json"

    if os.path.exists(archivo):
        try:
            with open(archivo, 'r', encoding='utf-8') as f:
                datos_usuario.update(json.load(f))
            return True
        except Exception:
            pass

    datos_usuario.update({
        "ingreso_mensual": 0, "metas": [], "presupuestos": {}, "ingresos": [], "gastos": []
    })
    return False

def guardar_perfil():
    if usuario_actual:
        with open(f"wallet_{usuario_actual}.json", 'w', encoding='utf-8') as f:
            json.dump(datos_usuario, f, indent=4)

# ==========================================
# ESTILOS VISUALES (PALETA WALL•ET)
# ==========================================
ESTILO_GLOBAL = """
<style>
    @import url('https://fonts.googleapis.com/css2?family=Segoe+UI:wght@400;600;700&display=swap');
    :root {
        --color-morado: #7E2885;
        --color-verde: #82BB28;
        --color-azul: #49BAC5;
        --color-fondo: #FEF0E7;
    }
    .wallet-container {
        font-family: 'Segoe UI', system-ui, sans-serif;
        background-color: var(--color-fondo);
        padding: 30px 20px;
        border-radius: 12px 12px 5px 5px;
        border-right: 4px dashed #a09a95;
        box-shadow: 0 8px 24px rgba(0, 0, 0, 0.08);
        max-width: 600px;
        margin: 0 auto;
    }
    .wallet-header { text-align: center; margin-bottom: 20px; }
    .wallet-title { color: var(--color-morado); font-size: 34px; font-weight: 700; margin-bottom: 5px; letter-spacing: -0.5px; }
    .wallet-subtitle { color: #4a5568; font-size: 18px; font-weight: 600; margin-top: 0; }
    .wallet-text { color: #718096; font-size: 15px; line-height: 1.5; }
    .wallet-card {
        background: #ffffff;
        border: 1px solid #e2e8f0;
        border-radius: 8px;
        padding: 20px;
        margin: 15px 0;
        box-shadow: 0 2px 8px rgba(0,0,0,0.04);
    }
</style>
"""

encabezado = widgets.HTML(ESTILO_GLOBAL + """
<div class="wallet-container">
    <div class="wallet-header">
        <h1 class="wallet-title">WALL•ET</h1>
        <h3 class="wallet-subtitle">Tu primer presupuesto</h3>
    </div>
</div>
""")

# Pantalla 1: Login
campo_usuario = widgets.Text(placeholder="Ej. Juan", layout=widgets.Layout(width="300px", margin="10px 0"))
boton_login = widgets.Button(description="Entrar", layout=widgets.Layout(width="200px", height="45px", margin="15px 0"))
boton_login.style.button_color = "#7E2885"
boton_login.style.text_color = "white"
mensaje_login = widgets.HTML("")

# Pantalla 2: Ingreso Inicial
campo_ingresos = widgets.FloatText(value=0, description="Ingresos ($):", style={"description_width": "initial"}, layout=widgets.Layout(width="300px", margin="10px 0"))
boton_continuar = widgets.Button(description="Empezar →", layout=widgets.Layout(width="200px", height="45px", margin="15px 0"))
boton_continuar.style.button_color = "#82BB28"
boton_continuar.style.text_color = "white"
mensaje_ingresos = widgets.HTML("")

def procesar_login(b):
    if not campo_usuario.value.strip():
        mensaje_login.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Por favor ingresa tu nombre.</p>"
        return

    existe = cargar_perfil(campo_usuario.value)
    if existe and datos_usuario["ingreso_mensual"] > 0:
        mostrar_menu()
    else:
        mostrar_bienvenida_ingresos()

def procesar_ingresos(b):
    if campo_ingresos.value <= 0:
        mensaje_ingresos.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Ingresa una cantidad mayor a cero.</p>"
    else:
        datos_usuario["ingreso_mensual"] = campo_ingresos.value
        guardar_perfil()
        mostrar_menu()

boton_login.on_click(procesar_login)
boton_continuar.on_click(procesar_ingresos)

def mostrar_bienvenida_ingresos():
    with pantalla:
        clear_output(wait=True)
        display(widgets.VBox([
            encabezado,
            widgets.HTML(f"<div style='text-align:center; color:#4a5568;'><h3 style='margin-bottom:5px;'>¡Hola, {usuario_actual.capitalize()}!</h3><p style='margin-top:0;'>¿Cuánto dinero recibes al mes?</p></div>"),
            campo_ingresos, boton_continuar, mensaje_ingresos
        ], layout=widgets.Layout(align_items="center", width="100%")))

def iniciar_app():
    with pantalla:
        clear_output(wait=True)
        display(widgets.VBox([
            encabezado,
            widgets.HTML("<div style='text-align:center; color:#4a5568;'><h3 style='margin-bottom:5px;'>Bienvenido</h3><p style='margin-top:0;'>Ingresa tu nombre para continuar</p></div>"),
            campo_usuario, boton_login, mensaje_login
        ], layout=widgets.Layout(align_items="center", width="100%")))
```

```python
#@title 🏠 WALL•ET | Menú principal

def crear_boton(texto, funcion, color=""):
    boton = widgets.Button(
        description=texto,
        layout=widgets.Layout(width="380px", height="55px", margin="6px 0")
    )
    if color == "morado":
        boton.style.button_color = "#7E2885"
        boton.style.text_color = "white"
    elif color == "verde":
        boton.style.button_color = "#82BB28"
        boton.style.text_color = "white"
    elif color == "azul":
        boton.style.button_color = "#49BAC5"
        boton.style.text_color = "white"

    boton.on_click(lambda b: funcion())
    return boton

def mostrar_menu():
    encabezado_menu = widgets.HTML(ESTILO_GLOBAL + f"""
    <div class="wallet-container" style="padding-bottom:10px;">
        <div class="wallet-header">
            <h1 class="wallet-title" style="font-size: 38px;">WALL•ET</h1>
            <h3 class="wallet-subtitle">Billetera de {usuario_actual.capitalize()}</h3>
            <p class="wallet-text">¿Qué quieres hacer hoy?</p>
        </div>
    </div>
    """)

    botones = [
        crear_boton("💵 Mi situación financiera", mostrar_situacion, "morado"),
        crear_boton("📝 Registrar movimientos", mostrar_movimientos, "verde"),
        crear_boton("🎯 Objetivos y ahorro", mostrar_objetivos_menu, "azul"),
        crear_boton("💰 Planificación financiera", mostrar_planificacion, "morado"),
        crear_boton("📋 Consultar mi información", mostrar_consultas, "verde"),
        crear_boton("📊 Mi información visual", mostrar_graficas, "azul"),
        crear_boton("🚪 Cambiar de usuario", iniciar_app)
    ]

    with pantalla:
        clear_output(wait=True)
        display(widgets.VBox(
            [encabezado_menu] + botones,
            layout=widgets.Layout(align_items="center", width="100%")
        ))
```

```python
#@title 💵 WALL•ET | Mi situación financiera

def mostrar_situacion():
    with pantalla:
        clear_output(wait=True)

        ingreso_mensual = datos_usuario["ingreso_mensual"]
        ingresos_extra = sum(i["monto"] for i in datos_usuario["ingresos"])
        gastos_registrados = sum(g["monto"] for g in datos_usuario["gastos"])
        ahorro_mensual = sum(m["ahorro_mensual"] for m in datos_usuario["metas"])
        ahorro_acumulado = sum(m["ahorrado"] for m in datos_usuario["metas"])

        dinero_restante = (ingreso_mensual + ingresos_extra) - ahorro_mensual - gastos_registrados

        encabezado_situacion = widgets.HTML(ESTILO_GLOBAL + f"""
        <div class="wallet-container">
            <div class="wallet-header">
                <h1 class="wallet-title">💵 Mi Billetera</h1>
                <h3 class="wallet-subtitle">Resumen de tu situación</h3>
            </div>

            <div style="display:grid; grid-template-columns:repeat(2, minmax(0, 1fr)); gap:16px; margin: 20px 0;">

                <div class="wallet-card" style="text-align:center; border-top: 8px solid var(--color-morado);">
                    <p style="color:#718096; margin:0; font-size:11px; text-transform:uppercase; font-weight:bold;">Control your money</p>
                    <p style="color:#4a5568; margin:5px 0 0 0; font-size:14px;">Ingresos Totales</p>
                    <h2 style="color:var(--color-morado); margin:8px 0 0 0;">{formato_dinero(ingreso_mensual + ingresos_extra)}</h2>
                </div>

                <div class="wallet-card" style="text-align:center; border-top: 8px solid var(--color-verde);">
                    <p style="color:#718096; margin:0; font-size:11px; text-transform:uppercase; font-weight:bold;">Know what you spend</p>
                    <p style="color:#4a5568; margin:5px 0 0 0; font-size:14px;">Gastos Realizados</p>
                    <h2 style="color:var(--color-verde); margin:8px 0 0 0;">{formato_dinero(gastos_registrados)}</h2>
                </div>

                <div class="wallet-card" style="text-align:center; border-top: 8px solid var(--color-azul); grid-column: span 2;">
                    <p style="color:#718096; margin:0; font-size:11px; text-transform:uppercase; font-weight:bold;">Learn how to save</p>
                    <p style="color:#4a5568; margin:5px 0 0 0; font-size:14px;">Ahorro Acumulado en Metas</p>
                    <h2 style="color:var(--color-azul); margin:8px 0 0 0;">{formato_dinero(ahorro_acumulado)}</h2>
                </div>

            </div>

            <div class="wallet-card" style="text-align:center; background-color:#ffffff; border: 2px solid var(--color-morado);">
                <p style="color:var(--color-morado); margin:0; font-weight:bold; text-transform:uppercase;">Dinero Disponible</p>
                <h1 style="color:var(--color-morado); font-size: 42px; margin: 10px 0;">{formato_dinero(dinero_restante)}</h1>
            </div>
        </div>
        """)

        boton_volver = widgets.Button(
            description="← Volver al menú",
            layout=widgets.Layout(width="220px", height="45px", margin="20px 0")
        )
        boton_volver.on_click(lambda b: mostrar_menu())

        display(widgets.VBox([
            encabezado_situacion,
            boton_volver
        ], layout=widgets.Layout(align_items="center", width="100%")))
```

```python
#@title ⚙️ WALL•ET | Módulos de la Aplicación

def mostrar_movimientos():
    with pantalla:
        clear_output(wait=True)
        encabezado = widgets.HTML(ESTILO_GLOBAL + """
        <div class="wallet-container">
            <div class="wallet-header">
                <p style="color:#718096; margin:0; font-size:12px; text-transform:uppercase; font-weight:bold;">Know what you spend</p>
                <h1 class="wallet-title" style="color:var(--color-verde);">📝 Movimientos</h1>
                <h3 class="wallet-subtitle">Registra un nuevo gasto o ingreso</h3>
            </div>
        </div>
        """)

        tipo = widgets.Dropdown(options=['Gasto', 'Ingreso Extra'], description='Tipo:', layout=widgets.Layout(width='300px', margin='10px 0'))
        concepto = widgets.Text(description='Concepto:', placeholder='Ej. Supermercado', layout=widgets.Layout(width='300px', margin='10px 0'))
        monto = widgets.FloatText(description='Monto ($):', layout=widgets.Layout(width='300px', margin='10px 0'))

        btn_guardar = widgets.Button(description='Guardar Movimiento', layout=widgets.Layout(width='200px', height='40px', margin='15px 0'))
        btn_guardar.style.button_color = '#82BB28'
        btn_guardar.style.text_color = 'white'
        mensaje = widgets.HTML("")

        def guardar(b):
            if monto.value <= 0 or concepto.value.strip() == "":
                mensaje.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Completa el concepto y un monto válido.</p>"
                return

            if tipo.value == 'Gasto':
                datos_usuario["gastos"].append({"concepto": concepto.value, "monto": monto.value})
                mensaje.value = f"<p style='color:var(--color-verde); text-align:center; font-weight:bold;'>¡Gasto de {formato_dinero(monto.value)} guardado!</p>"
            else:
                datos_usuario["ingresos"].append({"concepto": concepto.value, "monto": monto.value})
                mensaje.value = f"<p style='color:var(--color-morado); text-align:center; font-weight:bold;'>¡Ingreso de {formato_dinero(monto.value)} guardado!</p>"

            guardar_perfil()
            monto.value = 0
            concepto.value = ""

        btn_guardar.on_click(guardar)
        btn_volver = widgets.Button(description="← Volver al menú", layout=widgets.Layout(width="220px", height="45px", margin="20px 0"))
        btn_volver.on_click(lambda b: mostrar_menu())

        display(widgets.VBox([encabezado, tipo, concepto, monto, btn_guardar, mensaje, btn_volver], layout=widgets.Layout(align_items='center', width='100%')))


def mostrar_objetivos_menu():
    with pantalla:
        clear_output(wait=True)
        nombres_metas = [m["nombre"] for m in datos_usuario["metas"]]

        encabezado = widgets.HTML(ESTILO_GLOBAL + """
        <div class="wallet-container">
            <div class="wallet-header">
                <p style="color:#718096; margin:0; font-size:12px; text-transform:uppercase; font-weight:bold;">Learn how to save</p>
                <h1 class="wallet-title" style="color:var(--color-azul);">🎯 Objetivos</h1>
                <h3 class="wallet-subtitle">Aprende a ahorrar para tus metas</h3>
            </div>
        </div>
        """)

        titulo_crear = widgets.HTML("<h3 style='color:var(--color-azul); text-align:center; margin-bottom:5px;'>1. Crear Nueva Meta</h3>")
        nombre = widgets.Text(description='Meta:', placeholder='Ej. Viaje', layout=widgets.Layout(width='300px', margin='5px 0'))
        monto_total = widgets.FloatText(description='Costo total ($):', layout=widgets.Layout(width='300px', margin='5px 0'))
        ahorro_mensual = widgets.FloatText(description='Ahorro mes ($):', layout=widgets.Layout(width='300px', margin='5px 0'))

        btn_crear = widgets.Button(description='Crear Meta', layout=widgets.Layout(width='200px', height='40px', margin='15px 0'))
        btn_crear.style.button_color = '#49BAC5'
        btn_crear.style.text_color = 'white'

        titulo_aportar = widgets.HTML("<h3 style='color:var(--color-azul); text-align:center; margin-top:25px; margin-bottom:5px;'>2. Aportar a tus Metas</h3>")
        dropdown_metas = widgets.Dropdown(options=nombres_metas, description='Meta:', layout=widgets.Layout(width='300px', margin='5px 0'))
        monto_aporte = widgets.FloatText(description='Aporte ($):', layout=widgets.Layout(width='300px', margin='5px 0'))

        btn_aportar = widgets.Button(description='Registrar Aporte', layout=widgets.Layout(width='200px', height='40px', margin='15px 0'))
        btn_aportar.style.button_color = '#49BAC5'
        btn_aportar.style.text_color = 'white'

        mensaje = widgets.HTML("")

        def guardar_meta(b):
            if monto_total.value <= 0 or ahorro_mensual.value <= 0 or nombre.value.strip() == "":
                mensaje.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Verifica los datos ingresados.</p>"
                return
            datos_usuario["metas"].append({
                "nombre": nombre.value,
                "total": monto_total.value,
                "ahorro_mensual": ahorro_mensual.value,
                "ahorrado": 0
            })
            guardar_perfil()
            mostrar_objetivos_menu()

        def guardar_aporte(b):
            if not dropdown_metas.value or monto_aporte.value <= 0:
                mensaje.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Selecciona una meta y un monto válido.</p>"
                return
            for m in datos_usuario["metas"]:
                if m["nombre"] == dropdown_metas.value:
                    m["ahorrado"] += monto_aporte.value
                    guardar_perfil()
                    mensaje.value = f"<p style='color:var(--color-azul); text-align:center; font-weight:bold;'>¡Aporte de {formato_dinero(monto_aporte.value)} registrado en '{m['nombre']}'!</p>"
                    break
            monto_aporte.value = 0

        btn_crear.on_click(guardar_meta)
        btn_aportar.on_click(guardar_aporte)
        btn_volver = widgets.Button(description="← Volver al menú", layout=widgets.Layout(width="220px", height="45px", margin="20px 0"))
        btn_volver.on_click(lambda b: mostrar_menu())

        elementos = [encabezado, titulo_crear, nombre, monto_total, ahorro_mensual, btn_crear]
        if nombres_metas:
            elementos.extend([titulo_aportar, dropdown_metas, monto_aporte, btn_aportar])
        elementos.extend([mensaje, btn_volver])

        display(widgets.VBox(elementos, layout=widgets.Layout(align_items='center', width='100%')))


def mostrar_planificacion():
    with pantalla:
        clear_output(wait=True)
        encabezado = widgets.HTML(ESTILO_GLOBAL + """
        <div class="wallet-container">
            <div class="wallet-header">
                <p style="color:#718096; margin:0; font-size:12px; text-transform:uppercase; font-weight:bold;">Control your money</p>
                <h1 class="wallet-title" style="color:var(--color-morado);">💰 Planificación</h1>
                <h3 class="wallet-subtitle">Asigna topes máximos por categoría</h3>
            </div>
        </div>
        """)
        categoria = widgets.Text(description='Categoría:', placeholder='Ej. Comida', layout=widgets.Layout(width='300px', margin='10px 0'))
        limite = widgets.FloatText(description='Límite ($):', layout=widgets.Layout(width='300px', margin='10px 0'))

        btn_guardar = widgets.Button(description='Fijar Límite', layout=widgets.Layout(width='200px', height='40px', margin='15px 0'))
        btn_guardar.style.button_color = '#7E2885'
        btn_guardar.style.text_color = 'white'
        mensaje = widgets.HTML("")

        def guardar(b):
            if limite.value <= 0 or categoria.value.strip() == "":
                mensaje.value = "<p style='color:#e53e3e; text-align:center; font-weight:bold;'>Completa los campos correctamente.</p>"
                return
            datos_usuario["presupuestos"][categoria.value] = limite.value
            guardar_perfil()
            mensaje.value = f"<p style='color:var(--color-morado); text-align:center; font-weight:bold;'>Límite para {categoria.value} fijado en {formato_dinero(limite.value)}.</p>"
            categoria.value = ""
            limite.value = 0

        btn_guardar.on_click(guardar)
        btn_volver = widgets.Button(description="← Volver al menú", layout=widgets.Layout(width="220px", height="45px", margin="20px 0"))
        btn_volver.on_click(lambda b: mostrar_menu())
        display(widgets.VBox([encabezado, categoria, limite, btn_guardar, mensaje, btn_volver], layout=widgets.Layout(align_items='center', width='100%')))


def mostrar_consultas():
    with pantalla:
        clear_output(wait=True)

        html_content = ESTILO_GLOBAL + f"""
        <div class="wallet-container">
            <div class="wallet-header">
                <h1 class="wallet-title">📋 Mi Información</h1>
                <h3 class="wallet-subtitle">Historial de registros</h3>
            </div>

            <div class="wallet-card" style="border-left: 8px solid var(--color-verde);">
                <h3 style="color:var(--color-verde); margin-top:0;">Gastos Registrados</h3>
                <ul style="color:#4a5568;">
                    {''.join([f"<li>{g['concepto']}: <strong>{formato_dinero(g['monto'])}</strong></li>" for g in datos_usuario['gastos']]) or "<li>No hay gastos registrados.</li>"}
                </ul>
            </div>

            <div class="wallet-card" style="border-left: 8px solid var(--color-morado);">
                <h3 style="color:var(--color-morado); margin-top:0;">Ingresos Extra y Presupuestos</h3>
                <ul style="color:#4a5568;">
                    {''.join([f"<li>Ingreso - {i['concepto']}: <strong>{formato_dinero(i['monto'])}</strong></li>" for i in datos_usuario['ingresos']])}
                    {''.join([f"<li>Tope {cat}: <strong>{formato_dinero(lim)}</strong></li>" for cat, lim in datos_usuario['presupuestos'].items()])}
                    {("<li>Sin registros adicionales.</li>" if not datos_usuario['ingresos'] and not datos_usuario['presupuestos'] else "")}
                </ul>
            </div>

            <div class="wallet-card" style="border-left: 8px solid var(--color-azul);">
                <h3 style="color:var(--color-azul); margin-top:0;">Mis Metas</h3>
                <ul style="color:#4a5568;">
                    {''.join([f"<li>{m['nombre']}: <strong>{formato_dinero(m['ahorrado'])}</strong> de {formato_dinero(m['total'])}</li>" for m in datos_usuario['metas']]) or "<li>No hay metas registradas.</li>"}
                </ul>
            </div>
        </div>
        """

        btn_volver = widgets.Button(description="← Volver al menú", layout=widgets.Layout(width="220px", height="45px", margin="20px 0"))
        btn_volver.on_click(lambda b: mostrar_menu())

        display(widgets.VBox([widgets.HTML(html_content), btn_volver], layout=widgets.Layout(align_items='center', width='100%')))


def mostrar_graficas():
    with pantalla:
        clear_output(wait=True)

        ingreso_total = (datos_usuario["ingreso_mensual"] + sum(i["monto"] for i in datos_usuario["ingresos"])) or 1
        gastos_totales = sum(g["monto"] for g in datos_usuario["gastos"])
        porcentaje_gasto = min((gastos_totales / ingreso_total) * 100, 100)

        gastos_agrupados = {}
        for g in datos_usuario["gastos"]:
            concepto = g["concepto"].capitalize()
            gastos_agrupados[concepto] = gastos_agrupados.get(concepto, 0) + g["monto"]

        html_gastos = ""
        if not gastos_agrupados:
            html_gastos = "<p style='color:#718096; font-size:14px;'>Aún no hay gastos registrados.</p>"
        else:
            max_gasto = max(gastos_agrupados.values()) or 1
            for concepto, monto in sorted(gastos_agrupados.items(), key=lambda x: x[1], reverse=True)[:5]:
                pct = (monto / max_gasto) * 100
                html_gastos += f"""
                <div style="margin-bottom:12px;">
                    <div style="display:flex; justify-content:space-between; font-size:13px; color:#4a5568; margin-bottom:4px;">
                        <span>{concepto}</span>
                        <strong>{formato_dinero(monto)}</strong>
                    </div>
                    <div style="width:100%; background-color:#e2e8f0; border-radius:6px; height:10px; overflow:hidden;">
                        <div style="width:{pct}%; background-color:var(--color-verde); height:100%;"></div>
                    </div>
                </div>
                """

        html_metas = ""
        if not datos_usuario["metas"]:
            html_metas = "<p style='color:#718096; font-size:14px;'>Aún no hay metas registradas.</p>"
        else:
            for m in datos_usuario["metas"]:
                total = m["total"] or 1
                ahorrado = m["ahorrado"]
                pct_meta = min((ahorrado / total) * 100, 100)
                html_metas += f"""
                <div style="margin-bottom:12px;">
                    <div style="display:flex; justify-content:space-between; font-size:13px; color:#4a5568; margin-bottom:4px;">
                        <span>{m['nombre']}</span>
                        <span style="color:var(--color-azul); font-weight:bold;">{pct_meta:.1f}%</span>
                    </div>
                    <div style="width:100%; background-color:#e2e8f0; border-radius:6px; height:10px; overflow:hidden;">
                        <div style="width:{pct_meta}%; background-color:var(--color-azul); height:100%;"></div>
                    </div>
                </div>
                """

        html_content = ESTILO_GLOBAL + f"""
        <div class="wallet-container">
            <div class="wallet-header">
                <h1 class="wallet-title">📊 Visualiza tu dinero</h1>
                <h3 class="wallet-subtitle">Tus métricas clave</h3>
            </div>

            <div class="wallet-card" style="border: 1px solid var(--color-morado);">
                <p style="color:var(--color-morado); font-weight:bold; margin-bottom:10px; text-transform:uppercase; font-size:12px;">1. Gasto vs Ingresos Totales</p>
                <div style="width:100%; background-color:#e2e8f0; border-radius:10px; height:20px; overflow:hidden;">
                    <div style="width:{porcentaje_gasto}%; background-color:var(--color-morado); height:100%;"></div>
                </div>
                <p style="font-size:13px; color:#4a5568; margin-top:8px;">
                    Has consumido el <strong>{porcentaje_gasto:.1f}%</strong> de tu dinero disponible.
                </p>
            </div>

            <div class="wallet-card" style="border-top: 4px solid var(--color-verde);">
                <p style="color:var(--color-verde); font-weight:bold; margin-top:0; margin-bottom:15px; text-transform:uppercase; font-size:12px;">2. Tus Mayores Gastos</p>
                {html_gastos}
            </div>

            <div class="wallet-card" style="border-top: 4px solid var(--color-azul);">
                <p style="color:var(--color-azul); font-weight:bold; margin-top:0; margin-bottom:15px; text-transform:uppercase; font-size:12px;">3. Progreso de Ahorro</p>
                {html_metas}
            </div>
        </div>
        """

        btn_volver = widgets.Button(description="← Volver al menú", layout=widgets.Layout(width="220px", height="45px", margin="20px 0"))
        btn_volver.on_click(lambda b: mostrar_menu())

        display(widgets.VBox([widgets.HTML(html_content), btn_volver], layout=widgets.Layout(align_items='center', width='100%')))
```
